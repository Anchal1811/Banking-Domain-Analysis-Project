```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data
df = pd.read_csv(r"C:\Users\DELL\Downloads\Banking.csv")

# 2. Standardize column names (Remove hidden spaces)
df.columns = df.columns.str.strip()

# 3. Fix the Binning Logic for 'Estimated Income'
bins = [0, 100000, 300000, float('inf')]
labels = ['Low', 'Med', 'High']
df['Income Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=labels, right=False)

# 4. Corrected Categorical Columns List (Updated to match your CSV exactly)
categorical_cols = ["BRId", "GenderId", "IAId", "Amount of Credit Cards", 
                    "Nationality", "Occupation", "Fee Structure", "Loyalty Classification", "Income Band"]

# 5. Fixed Plotting Loop
for col in categorical_cols:
    if col in df.columns:
        print(f"Generating plot for: {col}")
        plt.figure(figsize=(10, 5))
        
        # If there are too many categories (like Occupation), show the top 10
        data_to_plot = df[col].value_counts()
        if len(data_to_plot) > 15:
            data_to_plot = data_to_plot.head(15)
            plt.title(f"Top 15 Categories in {col}")
        else:
            plt.title(f"Distribution of {col}")
            
        data_to_plot.plot(kind='bar', color='skyblue', edgecolor='black')
        plt.ylabel("Frequency")
        plt.xticks(rotation=45, ha='right')
        plt.tight_layout()
        plt.show()
    else:
        print(f"SKIPPED: '{col}' not found. Check if spelling matches: {df.columns.tolist()}")
```

```python
for i, predictor in enumerate(df[["BRId", "GenderId", "IAId", "Amount of Credit Cards", 
                                  "Nationality", "Occupation", "Fee Structure", 
                                  "Loyalty Classification", "Income Band"]]):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, hue='GenderId')
```

```python
# 1. Update your column list with the correct singular name
numerical_cols = [
    'Age', 'Estimated Income', 'Superannuation Savings', 
    'Credit Card Balance', 'Bank Loans', 'Bank Deposits', 
    'Checking Accounts', 'Saving Accounts', 
    'Foreign Currency Account', # Fixed: removed the 's'
    'Business Lending', 'Properties Owned', 'Risk Weighting'
]

# 2. Filter to only include columns that actually exist (Safety First!)
safe_cols = [c for c in numerical_cols if c in df.columns]

plt.figure(figsize=(15, 12))

for i, col in enumerate(safe_cols):
    # Ensure we don't exceed the 4x3 grid (12 slots)
    if i < 12:
        plt.subplot(4, 3, i + 1)
        sns.histplot(df[col], kde=True)
        plt.title(col)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# RE-DEFINE DF HERE TO BE SAFE
df = pd.read_csv("C:/Users/DELL/Downloads/Banking.csv")



cols_for_corr = [
    'Age', 'Estimated Income', 'Superannuation Savings', 
    'Amount of Credit Cards', 'Credit Card Balance', 'Bank Loans', 
    'Bank Deposits', 'Checking Accounts', 'Saving Accounts', 
    'Foreign Currency Account', 'Business Lending'
]

corr_matrix = df[cols_for_corr].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='crest', fmt=".2f")

plt.title('Correlation Matrix of Banking Variables')
plt.show()
```

```python
pairs_to_plot = [
    ('Bank Deposits', 'Saving Accounts'),
    ('Checking Accounts', 'Saving Accounts'),
    ('Checking Accounts', 'Foreign Currency Account'),
    ('Age', 'Superannuation Savings'),
    ('Estimated Income', 'Checking Accounts'),
    ('Bank Loans', 'Credit Card Balance'),
    ('Business Lending', 'Bank Loans'),
]


for x_col, y_col in pairs_to_plot:
    plt.figure(figsize=(8, 6))
    sns.regplot(
        data=df,
        x=x_col,
        y=y_col,
        scatter_kws={'alpha': 0.4},     # semi-transparent points
        line_kws={'color': 'red'}       # best-fit line color
    )
    plt.title(f'Relationship between {x_col} and {y_col}', fontsize=14)
    plt.xlabel(x_col, fontsize=12)
    plt.ylabel(y_col, fontsize=12)
    plt.tight_layout()
    plt.show()
```
